## Stage 1: Research Question & Objectives

### 1. Research Question
> *"Do amino acid substitutions in Mpox surface proteins (A29L/A35R) distinguish Clade I from Clade II lineages, and can machine learning accurately classify isolates using these sequence features?"*

### 2. Objectives
1. **Data Acquisition & QC:** Retrieve, clean, and document FASTA surface protein sequences (A29L / A35R) for Mpox Clade I and Clade II isolates from the NCBI Virus database.
2. **Preprocessing & EDA:** Analyze sequence length distributions, trim alignment gaps, and extract amino acid composition feature vectors across viral isolates.
3. **Core Alignment Analysis:** Compute pairwise distance matrices, perform Multiple Sequence Alignment (MSA), and construct phylogenetic trees mapping clade divergence.
4. **Machine Learning & Evaluation:** Train supervised classification models (KNN and Random Forest) on sequence feature vectors and evaluate performance against a dummy baseline classifier.

### 3. Hypothesis
Key amino acid substitutions alter sequence identity thresholds between Clade I and Clade II isolates, enabling supervised machine learning classifiers to distinguish clade origin with higher accuracy than dummy baseline models.
